# NLP Preprocessing Engine
This notebook builds a robust NLP preprocessing pipeline for handling noisy real-world text data.

## Task 1: Conceptual Understanding

### 1. Difference between "Love" and "love"
"Love" and "love" are treated as different tokens in case-sensitive NLP systems. Lowercasing helps normalize them.

### 2. What happens if stopwords are not removed?
Stopwords can add noise, increase dimensionality, and reduce model efficiency.

### 3. When removing stopwords is harmful
- Sentiment analysis ("not good")
- Question answering ("what is your name")

### 4. Stemming vs Lemmatization
Stemming truncates words (running → run), while lemmatization uses dictionary-based normalization.

```python
import re
from collections import Counter
```

```python
def preprocess_text(text):
    if not isinstance(text, str) or text.strip() == "":
        return []

    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"\S+@\S+", "", text)
    text = re.sub(r"\d+", "", text)
    text = re.sub(r'(.)\1+', r'\1', text)
    text = re.sub(r'[^a-z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()
    tokens = [word for word in tokens if len(word) > 2 or word in ['no', 'not']]

    return tokens
```

```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
```

```python
for text in sentences:
    tokens = preprocess_text(text)
    clean_sentence = " ".join(tokens)

    print("Original:", text)
    print("Tokens:", tokens)
    print("Cleaned:", clean_sentence)
    print("-" * 50)
```

```python
for text in sentences:
    tokens = preprocess_text(text)

    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_length = sum(len(word) for word in tokens) / total_tokens if total_tokens > 0 else 0

    print("Sentence:", text)
    print("Total Tokens:", total_tokens)
    print("Unique Tokens:", unique_tokens)
    print("Avg Token Length:", round(avg_length, 2))
    print("-" * 50)
```

## Analysis

Most noisy sentence: "Win $$$ now!!! Limited offer!!!"

Most meaningful after cleaning: "I am not happy with this"

```python
all_tokens = []

for text in sentences:
    all_tokens.extend(preprocess_text(text))

counter = Counter(all_tokens)

print("Top 10 frequent words:")
print(counter.most_common(10))

print("\nLeast 5 frequent words:")
print(counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(" ".join(tokens))

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
test_cases = ["", "😂😂😂😂", "123456"]

for case in test_cases:
    print("Input:", case)
    print("Output:", preprocess_text(case))
```
